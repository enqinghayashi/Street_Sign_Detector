# Create a directory with the new labels

```python
from pathlib import Path
import shutil

src_root = Path("./tcdata/labels")
dst_root = Path("./tcdata/labels1")

for split in ["mchar_train", "mchar_val"]:
    src_dir = src_root / split
    dst_dir = dst_root / split
    dst_dir.mkdir(parents=True, exist_ok=True)

    for txt in src_dir.glob("*.txt"):
        lines = txt.read_text().strip().splitlines()
        new_lines = []

        for line in lines:
            parts = line.split()
            if len(parts) == 5:
                parts[0] = "0"  # colapsar todas las clases
                new_lines.append(" ".join(parts))

        (dst_dir / txt.name).write_text("\n".join(new_lines))
```
