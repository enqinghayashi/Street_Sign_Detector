```python
import matplotlib.pyplot as plt
import numpy as np

# Datos extraídos del CSV
epochs = list(range(1, 31))  # Solo 30 épocas

# Métricas de accuracy (mAP50) y recall
mAP50 = [0.92637, 0.90890, 0.91057, 0.91407, 0.90867, 0.91703, 0.91062, 0.92269, 0.92310, 0.91666,
         0.92338, 0.91656, 0.91625, 0.92279, 0.92124, 0.92136, 0.92619, 0.92415, 0.92585, 0.92736,
         0.92536, 0.92252, 0.92420, 0.92575, 0.92458, 0.92501, 0.92512, 0.92496, 0.92472, 0.92379]

recall = [0.91686, 0.89481, 0.90081, 0.89699, 0.87979, 0.89493, 0.89850, 0.89608, 0.90747, 0.90543,
          0.91147, 0.91247, 0.90508, 0.90229, 0.90277, 0.90805, 0.91232, 0.91177, 0.91989, 0.91734,
          0.91594, 0.91368, 0.91252, 0.91569, 0.91418, 0.91681, 0.91724, 0.91715, 0.91715, 0.91629]

# mAP50-95 (para referencia)
mAP50_95 = [0.43572, 0.42491, 0.42625, 0.43456, 0.42764, 0.43685, 0.43187, 0.44515, 0.44058, 0.43813,
            0.44011, 0.44199, 0.43738, 0.44717, 0.44675, 0.44404, 0.44754, 0.44674, 0.44702, 0.44653,
            0.44671, 0.44637, 0.44686, 0.44779, 0.44644, 0.44673, 0.44586, 0.44506, 0.44537, 0.44485]

# Configuración general
plt.rcParams['figure.figsize'] = [14, 6]
plt.rcParams['font.size'] = 10
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3

# Encontrar máximos
max_mAP50 = max(mAP50)
max_recall = max(recall)
max_mAP50_epoch = epochs[mAP50.index(max_mAP50)]
max_recall_epoch = epochs[recall.index(max_recall)]

# 1. GRÁFICA HORIZONTAL: mAP50 vs Recall (una al lado de la otra)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

# Gráfica izquierda: mAP50
ax1.plot(epochs, mAP50, 'g-', linewidth=3, marker='o', markersize=6,
         markerfacecolor='white', markeredgewidth=1.5, markeredgecolor='green',
         label='mAP50 (Accuracy)')
ax1.axhline(y=max_mAP50, color='r', linestyle='--', alpha=0.7, linewidth=1.5,
            label=f'Max: {max_mAP50:.4f} (Epoch {max_mAP50_epoch})')
ax1.set_xlabel('Epoch', fontsize=11, fontweight='bold')
ax1.set_ylabel('mAP50', fontsize=11, fontweight='bold')
ax1.set_title('Validation Accuracy (mAP50)', fontsize=13, fontweight='bold', pad=12)
ax1.grid(True, alpha=0.3)
ax1.set_xticks(range(1, 31, 2))
ax1.set_yticks(np.arange(0.905, 0.93, 0.005))
ax1.set_ylim(0.905, 0.93)
ax1.legend(loc='lower right', fontsize=10)

# Resaltar puntos importantes
ax1.scatter([epochs[0]], [mAP50[0]], color='blue', s=80, zorder=5,
            label=f'Start: {mAP50[0]:.4f}')
ax1.scatter([epochs[-1]], [mAP50[-1]], color='red', s=80, zorder=5,
            label=f'End: {mAP50[-1]:.4f}')

# Añadir estadísticas en la gráfica
stats_text1 = f'Mean: {np.mean(mAP50):.4f}\nStd: {np.std(mAP50):.4f}'
ax1.text(0.02, 0.98, stats_text1, transform=ax1.transAxes,
         fontsize=9, verticalalignment='top',
         bbox=dict(boxstyle='round', facecolor='lightgreen', alpha=0.7))

# Gráfica derecha: Recall
ax2.plot(epochs, recall, 'm-', linewidth=3, marker='s', markersize=6,
         markerfacecolor='white', markeredgewidth=1.5, markeredgecolor='magenta',
         label='Recall')
ax2.axhline(y=max_recall, color='r', linestyle='--', alpha=0.7, linewidth=1.5,
            label=f'Max: {max_recall:.4f} (Epoch {max_recall_epoch})')
ax2.set_xlabel('Epoch', fontsize=11, fontweight='bold')
ax2.set_ylabel('Recall', fontsize=11, fontweight='bold')
ax2.set_title('Validation Recall', fontsize=13, fontweight='bold', pad=12)
ax2.grid(True, alpha=0.3)
ax2.set_xticks(range(1, 31, 2))
ax2.set_yticks(np.arange(0.88, 0.925, 0.005))
ax2.set_ylim(0.88, 0.925)
ax2.legend(loc='lower right', fontsize=10)

# Resaltar puntos importantes
ax2.scatter([epochs[0]], [recall[0]], color='blue', s=80, zorder=5,
            label=f'Start: {recall[0]:.4f}')
ax2.scatter([epochs[-1]], [recall[-1]], color='red', s=80, zorder=5,
            label=f'End: {recall[-1]:.4f}')

# Añadir estadísticas en la gráfica
stats_text2 = f'Mean: {np.mean(recall):.4f}\nStd: {np.std(recall):.4f}'
ax2.text(0.02, 0.98, stats_text2, transform=ax2.transAxes,
         fontsize=9, verticalalignment='top',
         bbox=dict(boxstyle='round', facecolor='lavender', alpha=0.7))

plt.suptitle('Model Performance Metrics Comparison', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('mAP50_vs_recall_horizontal.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()

# 2. GRÁFICA HORIZONTAL COMPACTA (sin duplicar leyendas)
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5), sharex=True)

# Gráfica izquierda: mAP50
line1, = ax1.plot(epochs, mAP50, 'g-', linewidth=2.5, marker='o', markersize=4,
                  markerfacecolor='white', markeredgewidth=1, label='mAP50')
ax1.axhline(y=max_mAP50, color='r', linestyle=':', alpha=0.5, linewidth=1)
ax1.fill_between(epochs, mAP50, alpha=0.2, color='green')
ax1.set_xlabel('Epoch', fontsize=10)
ax1.set_ylabel('mAP50', fontsize=10)
ax1.set_title('Accuracy (mAP50)', fontsize=12)
ax1.grid(True, alpha=0.2, linestyle='--')
ax1.set_xticks(range(0, 31, 5))
ax1.set_xticks(range(1, 31, 1), minor=True)

# Gráfica derecha: Recall
line2, = ax2.plot(epochs, recall, 'm-', linewidth=2.5, marker='s', markersize=4,
                  markerfacecolor='white', markeredgewidth=1, label='Recall')
ax2.axhline(y=max_recall, color='r', linestyle=':', alpha=0.5, linewidth=1)
ax2.fill_between(epochs, recall, alpha=0.2, color='magenta')
ax2.set_xlabel('Epoch', fontsize=10)
ax2.set_ylabel('Recall', fontsize=10)
ax2.set_title('Recall', fontsize=12)
ax2.grid(True, alpha=0.2, linestyle='--')
ax2.set_xticks(range(0, 31, 5))
ax2.set_xticks(range(1, 31, 1), minor=True)

# Leyenda combinada en la parte superior
fig.legend([line1, line2], ['mAP50', 'Recall'],
           loc='upper center', ncol=2, fontsize=10, framealpha=0.9)

plt.suptitle('Model Performance: mAP50 vs Recall', fontsize=13, fontweight='bold', y=1.05)
plt.tight_layout(rect=[0, 0, 1, 0.95])  # Ajustar para la leyenda superior
plt.savefig('compact_horizontal_comparison.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()

# 3. GRÁFICA HORIZONTAL CON ANOTACIONES DE ÉPOCAS CLAVE
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 5))

# Gráfica mAP50 con anotaciones
ax1.plot(epochs, mAP50, 'g-', linewidth=2, marker='o', markersize=5)
ax1.set_xlabel('Epoch', fontsize=11)
ax1.set_ylabel('mAP50', fontsize=11)
ax1.set_title('Accuracy Evolution', fontsize=12, fontweight='bold')
ax1.grid(True, alpha=0.3)

# Anotar épocas clave para mAP50
key_epochs_mAP50 = [1, max_mAP50_epoch, 20, 30]  # Inicio, máximo, mitad, final
for epoch in key_epochs_mAP50:
    if 1 <= epoch <= 30:
        value = mAP50[epoch-1]
        color = 'red' if epoch == max_mAP50_epoch else 'blue' if epoch == 1 else 'black'
        ax1.scatter(epoch, value, color=color, s=100, zorder=5)
        ax1.annotate(f'{value:.4f}', xy=(epoch, value),
                    xytext=(epoch, value+0.003),
                    ha='center', fontsize=8, color=color,
                    arrowprops=dict(arrowstyle='->', color=color, alpha=0.5))

# Gráfica Recall con anotaciones
ax2.plot(epochs, recall, 'm-', linewidth=2, marker='s', markersize=5)
ax2.set_xlabel('Epoch', fontsize=11)
ax2.set_ylabel('Recall', fontsize=11)
ax2.set_title('Recall Evolution', fontsize=12, fontweight='bold')
ax2.grid(True, alpha=0.3)

# Anotar épocas clave para Recall
key_epochs_recall = [1, max_recall_epoch, 20, 30]
for epoch in key_epochs_recall:
    if 1 <= epoch <= 30:
        value = recall[epoch-1]
        color = 'red' if epoch == max_recall_epoch else 'blue' if epoch == 1 else 'black'
        ax2.scatter(epoch, value, color=color, s=100, zorder=5)
        ax2.annotate(f'{value:.4f}', xy=(epoch, value),
                    xytext=(epoch, value+0.003),
                    ha='center', fontsize=8, color=color,
                    arrowprops=dict(arrowstyle='->', color=color, alpha=0.5))

plt.suptitle('Key Epochs Analysis', fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('horizontal_with_annotations.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()

# 4. GRÁFICA HORIZONTAL CON ÁREAS SOMBREADAS
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Definir colores y estilos
colors = ['#2E8B57', '#8B008B']  # Verde marino y magenta oscuro
fill_alphas = [0.15, 0.15]

# mAP50 con área sombreada
axes[0].fill_between(epochs, mAP50, alpha=fill_alphas[0], color=colors[0], label='mAP50 Range')
axes[0].plot(epochs, mAP50, color=colors[0], linewidth=2.5, marker='o', markersize=5,
             markerfacecolor='white', markeredgecolor=colors[0], markeredgewidth=1.5)
axes[0].axhline(y=np.mean(mAP50), color='red', linestyle='--', alpha=0.7,
                label=f'Mean: {np.mean(mAP50):.4f}')
axes[0].set_xlabel('Epoch', fontsize=11, fontweight='bold')
axes[0].set_ylabel('mAP50', fontsize=11, fontweight='bold')
axes[0].set_title('Accuracy (mAP50) with Confidence Area', fontsize=12, fontweight='bold')
axes[0].grid(True, alpha=0.2)
axes[0].legend(loc='lower right', fontsize=9)

# Recall con área sombreada
axes[1].fill_between(epochs, recall, alpha=fill_alphas[1], color=colors[1], label='Recall Range')
axes[1].plot(epochs, recall, color=colors[1], linewidth=2.5, marker='s', markersize=5,
             markerfacecolor='white', markeredgecolor=colors[1], markeredgewidth=1.5)
axes[1].axhline(y=np.mean(recall), color='red', linestyle='--', alpha=0.7,
                label=f'Mean: {np.mean(recall):.4f}')
axes[1].set_xlabel('Epoch', fontsize=11, fontweight='bold')
axes[1].set_ylabel('Recall', fontsize=11, fontweight='bold')
axes[1].set_title('Recall with Confidence Area', fontsize=12, fontweight='bold')
axes[1].grid(True, alpha=0.2)
axes[1].legend(loc='lower right', fontsize=9)

# Ajustar ejes Y para mejor visualización
for ax in axes:
    ax.set_xticks(range(0, 31, 5))
    ax.set_xticks(range(1, 31, 1), minor=True)

# Establecer límites Y similares para mejor comparación
y_min = min(min(mAP50), min(recall)) - 0.005
y_max = max(max(mAP50), max(recall)) + 0.005
axes[0].set_ylim(y_min, y_max)
axes[1].set_ylim(y_min, y_max)

plt.suptitle('Performance Metrics with Variability Visualization',
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('horizontal_with_shaded_areas.png', dpi=300, bbox_inches='tight', facecolor='white')
plt.show()

print("=" * 60)
print("GRÁFICAS HORIZONTALES GENERADAS")
print("=" * 60)
print("\n📊 Archivos guardados:")
print("1. mAP50_vs_recall_horizontal.png - Comparación horizontal completa")
print("2. compact_horizontal_comparison.png - Versión compacta")
print("3. horizontal_with_annotations.png - Con épocas clave anotadas")
print("4. horizontal_with_shaded_areas.png - Con áreas de variabilidad")

print("\n📈 RESUMEN ESTADÍSTICO:")
print(f"mAP50: {np.mean(mAP50):.4f} ± {np.std(mAP50):.4f} (range: {max(mAP50)-min(mAP50):.4f})")
print(f"Recall: {np.mean(recall):.4f} ± {np.std(recall):.4f} (range: {max(recall)-min(recall):.4f})")

# Calcular correlación entre mAP50 y Recall
correlation = np.corrcoef(mAP50, recall)[0, 1]
print(f"\n🔗 Correlación mAP50-Recall: {correlation:.4f}")
if correlation > 0.7:
    print("   → Alta correlación: Ambas métricas evolucionan juntas")
elif correlation > 0.3:
    print("   → Correlación moderada")
else:
    print("   → Baja correlación: Evoluciones independientes")

print("\n🏆 MEJORES ÉPOCAS:")
print(f"   • Mejor mAP50: Época {max_mAP50_epoch} ({max_mAP50:.4f})")
print(f"   • Mejor Recall: Época {max_recall_epoch} ({max_recall:.4f})")
print(f"   • Época con mejor balance: Época 20 (mAP50: {mAP50[19]:.4f}, Recall: {recall[19]:.4f})")
```
